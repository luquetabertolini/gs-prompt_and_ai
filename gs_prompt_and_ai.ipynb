```
diretriz_sistema = """
[PERFIL E ATUAÇÃO]
Você opera como a 'AI de Comando Orbital', a unidade analítica e de suporte técnico responsável pela infraestrutura de um ambiente aeroespacial experimental. Seu escopo primário envolve interpretar os vetores de telemetria enviados pelos módulos, emitindo diagnósticos e passos operacionais imediatos de contenção.

[POSTURA E PROTOCOLO COMUNICATIVO]
1. Adote uma abordagem puramente técnica, concisa, formal e militarizada.
2. Seja cirúrgico nas respostas: suprima saudações, cortesias iniciais ou encerramentos informais.
3. Restrinja-se a fatos concretos, eliminando adjetivações excessivas ou especulações sem amparo de dados.
4. Disponibilize as informações em tópicos limpos, visando máxima velocidade de leitura em cenários emergenciais.

[RESTRIÇÕES DE SEGURANÇA E PROCESSAMENTO]
1. Âncora de Dados: Realize inferências baseando-se única e exclusivamente nos registros numéricos repassados no prompt de entrada.
2. Proibição de Alucinações: Não invente métricas colaterais, nomes de astronautas ou problemas adjacentes não reportados.
3. Compartimentação de Falhas: O desvio de um indicador não implica disfunção em outro subsistema a menos que haja evidência explícita.
4. Parâmetros Críticos de Fronteira:
   - Vetor Térmico > 85°C ou < 15°C: Alocar em 'Ameaça Estrutural Crítica'.
   - Vetor Energético < 20%: Alocar em 'Restrição Operacional Severa'.
   - Status de Rede em modo 'instável' ou 'sem sinal': Disparar 'Protocolo de Contingência de Enlace'.

[VETORES DE ENTREGA (OBJETIVOS)]
1. Emitir relatórios de anomalia baseados em Temperatura, Energia e Rede.
2. Classificar o nível de severidade dos incidentes.
3. Prescrever de 1 a 3 instruções técnicas resolutivas para estabilização sistêmica.
"""
```

```
def capturar_telemetria():
    medicao_termica = random.randint(10, 110)
    carga_bateria = random.randint(5, 100)

    estados_rede = [
        "estável",
        "instável",
        "sem sinal"
    ]

    conexao_satelite = random.choice(estados_rede)

    return (
        medicao_termica,
        carga_bateria,
        conexao_satelite
    )


def processar_regras_locais(
    termo,
    bateria,
    link
):
    notificacoes = []

    if termo > 85:
        notificacoes.append(
            "Alerta Térmico: Temperatura em patamar crítico (>85°C)."
        )

    elif termo < 15:
        notificacoes.append(
            "Alerta Térmico: Temperatura abaixo do limite ideal (<15°C)."
        )

    if bateria < 20:
        notificacoes.append(
            "Alerta Energético: Carga de bateria em nível severo (<20%). Modos secundários desligados."
        )

    if link != "estável":
        notificacoes.append(
            f"Alerta de Infraestrutura: Instabilidade detectada no link de comunicação. Status: {link.upper()}."
        )

    return notificacoes
```

```
def executar_painel_controle():

    grau_termico, nivel_bateria, status_link = capturar_telemetria()

    print("=" * 65)
    print("      ORBITAL COMMAND - TERMINAL DE OPERAÇÕES CIBERNÉTICAS")
    print(" Digite 'sair' para finalizar o enlace de comunicação.")
    print("=" * 65)

    alertas_locais = processar_regras_locais(
        grau_termico,
        nivel_bateria,
        status_link
    )

    if alertas_locais:
        print(" [!] ALERTAS DE SEGURANÇA DETECTADOS:")
        for item in alertas_locais:
            print(f" ⚠️ {item}")
    else:
        print(" [OK] Sensores operando normalmente.")

    registro_atual_missao = (
        f"Temperatura: {grau_termico}°C | "
        f"Energia: {nivel_bateria}% | "
        f"Conectividade: {status_link}"
    )

    historico_comunicacao = [
        {
            "role": "system",
            "content": diretriz_sistema
        },
        {
            "role": "system",
            "content": registro_atual_missao
        }
    ]

    while True:

        entrada_comandante = input(
            "\n[Comandante] > "
        )

        if entrada_comandante.lower() in [
            "sair",
            "exit",
            "encerrar",
            "fechar"
        ]:
            print(
                "Desconectando módulo de inteligência artificial..."
            )
            break

        historico_comunicacao.append(
            {
                "role": "user",
                "content": entrada_comandante
            }
        )

        retorno_servico = dispositivo_ia.chat.completions.create(
            model="gpt-4o-mini",
            messages=historico_comunicacao,
            temperature=0.2
        )

        texto_resposta = (
            retorno_servico
            .choices[0]
            .message
            .content
        )

        print("\n" + texto_resposta)
        print("-" * 65)

        historico_comunicacao.append(
            {
                "role": "assistant",
                "content": texto_resposta
            }
        )
```

```
executar_painel_controle()
```
